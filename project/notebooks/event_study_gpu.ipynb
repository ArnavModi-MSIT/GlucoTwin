```python
from pathlib import Path
import os,sys,subprocess,json
ROOT=Path(os.environ.get('GLUCOTWIN_ROOT','/content/GlucoTwin/project'))
# Kaggle users can set ROOT=Path('/kaggle/working/GlucoTwin/project').
assert (ROOT/'scripts/run_event_study.py').exists(), 'Place the current project here first'
assert (ROOT/'configs/cgmacros_splits.json').exists(), 'Prepare the local manifest first'
```

The runner checks the fitted XGBoost device and fails if CUDA silently falls back to CPU. CPU and CUDA results may differ numerically. Results are separate development experiments, not medical validation. Never lower the gates to obtain a passing result.


```python
subprocess.run([sys.executable,str(ROOT/'scripts/run_event_study.py'),'--study','gpu_v2','--device','cuda'],cwd=ROOT,check=True)
report=json.loads((ROOT/'reports/event_study_gpu_v2.json').read_text())
assert report['gpu_execution_verified']
print(json.dumps({'pooled':report['pooled'],'gates':report['pooled_engineering_gate_passed'],'versions':report['versions']},indent=2))
```
